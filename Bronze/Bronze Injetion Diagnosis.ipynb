## Using Autoloader batch Mode we will load all the source files from the ADLS storage

### Declaring all the paths for Diagnosis raw data

```python
source_path = '/Volumes/healthcare/bronze/bronze-data/diagnosis/'
checkpoint_path = '/Volumes/healthcare/bronze/checkpoint/bronze_diagnosis_files/v1'
schema_path = '/Volumes/healthcare/bronze/schematracking/diagnosis_schema'
```

```python
autoloader_df = (
                    spark.readStream
                    .format('cloudFiles')
                    .option('cloudFiles.format','csv')
                    .option('header','true')
                    .option('cloudFiles.schemaLocation',schema_path)
                    .option('cloudFiles.maxFilesPerTrigger','1')
                    .load(source_path) 
                    )
```

```python
(
    autoloader_df.drop('_rescued_data')
    .writeStream
    .format('delta')
    .outputMode('append')
    .option('checkpointLocation',checkpoint_path)
    .trigger(availableNow=True)
    .toTable('healthcare.bronze.diagnosis')
)
```
