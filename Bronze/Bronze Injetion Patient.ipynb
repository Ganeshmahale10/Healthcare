### Declaring all the paths for Patient raw data

```python
source_path = '/Volumes/healthcare/bronze/bronze-data/patient/'
checkpoint_path = '/Volumes/healthcare/bronze/checkpoint/bronze_patient_files/v1'
schema_path = '/Volumes/healthcare/bronze/schematracking/patient_schema'
```

```python
autoloader_df = (	
		spark.readStream
		.format('cloudFiles')
		.option('cloudFiles.format','csv')
		.option('header','true')
		.option('cloudFiles.schemaLocation',schema_path)
		.option('cloudFiles.maxFilesPerTrigger','1')
		.load(source_path)
		)

	
(
	autoloader_df.drop('_rescued_data')
	.writeStream
	.format('delta')
	.outputMode('append')
	.option('checkpointLocation',checkpoint_path)
	.trigger(availableNow = True)
	.toTable('healthcare.bronze.patient')
)
```
